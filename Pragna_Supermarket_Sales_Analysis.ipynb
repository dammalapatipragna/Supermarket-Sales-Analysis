````python
# ==========================================
# SUPERMARKET SALES DATA ANALYSIS
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", palette="muted")


# ==========================================
# 1. LOAD DATASET
# ==========================================

# Updated with correct absolute path
df = pd.read_csv('/content/SUPER MARKET DATA - supermarket_sales_500_rows.csv')

print("=== First 5 Rows ===")
display(df.head())


# ==========================================
# 2. BASIC INFORMATION
# ==========================================

print("\n=== Dataset Shape ===")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


print("\n=== Column Names ===")
print(df.columns.tolist())


print("\n=== Data Information ===")
df.info()


# ==========================================
# 3. DESCRIPTIVE STATISTICS
# ==========================================

print("\n=== Descriptive Statistics ===")
display(df.describe(include='all'))


# ==========================================
# 4. CHECK MISSING VALUES
# ==========================================

print("\n=== Missing Values ===")
print(df.isnull().sum())


# ==========================================
# 5. CHECK DUPLICATE VALUES
# ==========================================

print("\n=== Duplicate Rows ===")
print(df.duplicated().sum())


# ==========================================
# 6. DATE CONVERSION
# ==========================================

df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month
df['DayOfWeek'] = df['Date'].dt.day_name()


print("\n=== Date Range ===")
print("Start Date:", df['Date'].min())
print("End Date:", df['Date'].max())


# ==========================================
# 7. BASIC BUSINESS SUMMARY
# ==========================================

print("\n=== Business Summary ===")

print("Total Sales: ₹", round(df['Sales'].sum(), 2))
print("Total Quantity Sold:", df['Quantity'].sum())
print("Average Sales per Transaction: ₹", round(df['Sales'].mean(), 2))
print("Average Customer Rating:", round(df['Rating'].mean(), 2))


# ==========================================
# 8. SALES DISTRIBUTION
# ==========================================

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)

sns.histplot(
    df['Sales'],
    bins=30,
    kde=True
)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")


plt.subplot(2, 2, 2)

sns.boxplot(
    x=df['Sales']
)

plt.title("Boxplot of Sales")
plt.xlabel("Sales")


plt.subplot(2, 2, 3)

sns.histplot(
    df['Quantity'],
    bins=10,
    kde=True
)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")


plt.subplot(2, 2, 4)

sns.histplot(
    df['Rating'],
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Rating")
plt.xlabel("Rating")
plt.ylabel("Frequency")


plt.tight_layout()
plt.show()


# ==========================================
# 9. SALES BY PRODUCT
# ==========================================

sales_by_product = (
    df.groupby('Product')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\n=== Sales by Product ===")
display(sales_by_product)


plt.figure(figsize=(12, 6))

sns.barplot(
    x=sales_by_product.index,
    y=sales_by_product.values
)

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=60)

plt.tight_layout()
plt.show()


# ==========================================
# 10. SALES BY CITY
# ==========================================

sales_by_city = (
    df.groupby('City')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\n=== Sales by City ===")
display(sales_by_city)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=sales_by_city.index,
    y=sales_by_city.values
)

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 11. SALES BY BRANCH
# ==========================================

sales_by_branch = (
    df.groupby(['Branch', 'City'])['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\n=== Sales by Branch ===")
display(sales_by_branch)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=df['Branch'],
    y=df['Sales']
)

plt.title("Sales Distribution by Branch")
plt.xlabel("Branch")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 12. SALES BY CATEGORY
# ==========================================

sales_by_category = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\n=== Sales by Category ===")
display(sales_by_category)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=sales_by_category.index,
    y=sales_by_category.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ==========================================
# 13. PAYMENT METHOD ANALYSIS
# ==========================================

payment_counts = df['Payment'].value_counts()

print("\n=== Payment Method Usage ===")
display(payment_counts)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=payment_counts.index,
    y=payment_counts.values
)

plt.title("Most Popular Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()


# ==========================================
# 14. CUSTOMER TYPE ANALYSIS
# ==========================================

sales_by_customer = (
    df.groupby('Customer Type')['Sales']
    .mean()
    .sort_values(ascending=False)
)

print("\n=== Average Sales by Customer Type ===")
display(sales_by_customer)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=sales_by_customer.index,
    y=sales_by_customer.values
)

plt.title("Average Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Average Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 15. GENDER-WISE SALES
# ==========================================

sales_by_gender = (
    df.groupby('Gender')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\n=== Sales by Gender ===")
display(sales_by_gender)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=sales_by_gender.index,
    y=sales_by_gender.values
)

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 16. QUANTITY VS SALES
# ==========================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    x='Quantity',
    y='Sales',
    data=df,
    alpha=0.6
)

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 17. UNIT PRICE VS SALES
# ==========================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    x='Unit Price',
    y='Sales',
    data=df,
    alpha=0.6
)

plt.title("Unit Price vs Sales")
plt.xlabel("Unit Price")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()


# ==========================================
# 18. SALES BY DAY OF WEEK
# ==========================================

sales_by_day = (
    df.groupby('DayOfWeek')['Sales']
    .sum()
)

day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

sales_by_day = sales_by_day.reindex(day_order)

print("\n=== Sales by Day of Week ===")
display(sales_by_day)


plt.figure(figsize=(10, 5))

sns.barplot(
    x=sales_by_day.index,
    y=sales_by_day.values
)

plt.title("Total Sales by Day of Week")
plt.xlabel("Day")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ==========================================
# 19. CORRELATION HEATMAP
# ==========================================

numeric_df = df.select_dtypes(include=[np.number])

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numeric Features")

plt.tight_layout()
plt.show()


# ==========================================
# 20. IMPORTANT BUSINESS INSIGHTS
# ==========================================

highest_product = sales_by_product.idxmax()
highest_product_sales = sales_by_product.max()

best_city = sales_by_city.idxmax()
best_city_sales = sales_by_city.max()

best_branch = (
    df.groupby(['Branch', 'City'])['Sales']
    .sum()
    .idxmax()
)

best_branch_sales = (
    df.groupby(['Branch', 'City'])['Sales']
    .sum()
    .max()
)

highest_category = sales_by_category.idxmax()
highest_category_sales = sales_by_category.max()

most_used_payment = payment_counts.idxmax()
most_used_payment_count = payment_counts.max()

highest_customer_type = sales_by_customer.idxmax()
highest_customer_average = sales_by_customer.max()


print("\n========================================")
print("          FINAL BUSINESS INSIGHTS")
print("========================================")

print(
    f"\n1. Highest-selling product: "
    f"{highest_product} (₹{highest_product_sales:,.2f})"
)

print(
    f"\n2. Best-performing city: "
    f"{best_city} (₹{best_city_sales:,.2f})"
)

print(
    f"\n3. Best-performing branch: "
    f"Branch {best_branch[0]} ({best_branch[1]}) "
    f"(₹{best_branch_sales:,.2f})"
)

print(
    f"\n4. Highest-selling category: "
    f"{highest_category} (₹{highest_category_sales:,.2f})"
)

print(
    f"\n5. Most popular payment method: "
    f"{most_used_payment} ({most_used_payment_count} transactions)"
)

print(
    f"\n6. Customer type with higher average spending: "
    f"{highest_customer_type} (₹{highest_customer_average:,.2f})"
)

print(
    f"\n7. Average customer rating: "
    f"{df['Rating'].mean():.2f}/5"
)

print(
    f"\n8. Total sales: "
    f"₹{df['Sales'].sum():,.2f}"
)

print(
    f"\n9. Total quantity sold: "
    f"{df['Quantity'].sum()}"
)

print("\n========================================")
````
